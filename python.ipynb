## 6.3 Sets
A set in Python is an unordered, mutable collection of unique elements.
It does not allow duplicates and does not maintain order.

| Method                              | Description                               | Application                      | Example (Input → Output)                                  |
| ----------------------------------- | ----------------------------------------- | -------------------------------- | --------------------------------------------------------- |
| **add(x)**                          | Adds element `x` to the set               | Insert new item                  | `{1,2}.add(3)` → `{1,2,3}`                                |
| **clear()**                         | Removes all elements                      | Reset set                        | `{1,2}.clear()` → `set()`                                 |
| **copy()**                          | Returns shallow copy                      | Duplicate set                    | `s.copy()` → `{…}`                                        |
| **difference(s2)**                  | Elements in s1 but not in s2              | Find non-overlapping items       | `{1,2,3}.difference({2,3})` → `{1}`                       |
| **difference_update(s2)**           | Removes elements of s2 from original set  | In-place difference              | `s={1,2,3}; s.difference_update({2}); s` → `{1,3}`        |
| **discard(x)**                      | Removes `x` if exists (no error)          | Safe removal                     | `{1,2}.discard(2)` → `{1}`                                |
| **intersection(s2)**                | Elements common in both sets              | Find shared items                | `{1,2,3}.intersection({2,3,4})` → `{2,3}`                 |
| **intersection_update(s2)**         | Keeps only common elements (in-place)     | In-place intersection            | `s={1,2,3}; s.intersection_update({2,3})` → `{2,3}`       |
| **isdisjoint(s2)**                  | True if no common elements                | Check if two sets overlap        | `{1,2}.isdisjoint({3})` → `True`                          |
| **issubset(s2)**                    | Checks if set is subset of s2             | Check containment                | `{1,2}.issubset({1,2,3})` → `True`                        |
| **issuperset(s2)**                  | Checks if set is superset of s2           | Check if s contains all elements | `{1,2,3}.issuperset({1})` → `True`                        |
| **pop()**                           | Removes and returns random element        | Remove unknown element           | `{1,2,3}.pop()` → e.g. `1`                                |
| **remove(x)**                       | Removes `x`; raises error if absent       | Strict removal                   | `{1,2}.remove(2)` → `{1}`                                 |
| **symmetric_difference(s2)**        | Returns elements in either set, not both  | XOR set operation                | `{1,2}.symmetric_difference({2,3})` → `{1,3}`             |
| **symmetric_difference_update(s2)** | Updates set with symmetric difference     | In-place XOR                     | `s={1,2}; s.symmetric_difference_update({2,3})` → `{1,3}` |
| **union(s2,... )**                  | Returns all unique elements from all sets | Combine many sets                | `{1,2}.union({2,3},{3,4})` → `{1,2,3,4}`                  |
| **update(s2)**                      | Adds all elements of s2                   | In-place merge                   | `s={1,2}; s.update({2,3})` → `{1,2,3}`                    |

## Dictionary :
Dictionary is key-value pair where we can access value with help of key.

| Method                        | Description                                                    | Typical Application                       | Example (Input → Output)                                         |
| ----------------------------- | -------------------------------------------------------------- | ----------------------------------------- | ---------------------------------------------------------------- |
| **clear()**                   | Removes all items from the dictionary                          | Reset a dictionary to empty               | `d = {'a':1,'b':2}; d.clear(); print(d)` → `{}`                  |
| **copy()**                    | Returns a shallow copy of the dictionary                       | Duplicate dict without linking references | `d={'x':1}; c=d.copy(); print(c)` → `{'x':1}`                    |
| **fromkeys(seq, value=None)** | Creates a new dict with keys from sequence and one value       | Initialize dictionary with default values | `d=dict.fromkeys(['a','b'],0); print(d)` → `{'a':0,'b':0}`       |
| **get(key, default=None)**    | Returns value for key; returns default if key not found        | Safe key access without error             | `d={'a':1}; print(d.get('b',0))` → `0`                           |
| **items()**                   | Returns a view object of `(key, value)` pairs                  | Iterating through items                   | `d={'a':1}; print(list(d.items()))` → `[('a',1)]`                |
| **keys()**                    | Returns a view of dictionary keys                              | Loop through keys                         | `d={'a':1}; print(list(d.keys()))` → `['a']`                     |
| **pop(key, default)**         | Removes key and returns its value; returns default if missing  | Remove and get item at same time          | `d={'a':1}; print(d.pop('a'))` → `1`                             |
| **popitem()**                 | Removes and returns last inserted `(key, value)` pair          | LIFO removal (stack-like)                 | `d={'a':1,'b':2}; print(d.popitem())` → `('b',2)`                |
| **setdefault(key, default)**  | Returns value of key; inserts key with default if missing      | Insert if absent, otherwise get           | `d={}; print(d.setdefault('a',10)); print(d)` → `10`, `{'a':10}` |
| **update(other_dict)**        | Updates the dictionary with other dictionary or iterable pairs | Merge dictionaries                        | `d={'a':1}; d.update({'b':2}); print(d)` → `{'a':1,'b':2}`       |
| **values()**                  | Returns a view of dictionary values                            | Loop through values                       | `d={'a':1}; print(list(d.values()))` → `[1]`                     |

## Recursive Function
A function that calls itself in order to solve a problem. It must have a base case to terminate the recursion.

```python
def fact_num(n):
    print("number_check",n)
    if n==1:
        return 1
    elif n==0:
        return 0
    else:
        return n*fact_num(n-1)
print(fact_num(10))
```

Hiding internal implemention and showing only what is necessary to show user.

### Inheritance
Inheritance means inherit the properties from another class

#### Types Of Inheritance:
1. Single Inheritance : It means single child class inherit properties from single parent class.
2. Multiple Inheritance: It means single child class inherit properties from multiple parent class.
3. Multilevel Inheritance: It means child class inherit properties from parent class which inherited from his parent class.
4. Hierarchical Inheritance: It Means multiple child class inheriting properties from single parent class.
5. Hybrid Inheritance : A combination of two or more inheritance types (e.g., multilevel + multiple).

| **Type of Inheritance**                            | **Description**                                                                                                            | **Code Example**                                                                                                                                                                              | **Application / Use Case**                                                                       |
| -------------------------------------------------- | -------------------------------------------------------------------------------------------------------------------------- | --------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------ |
| **1. Single Inheritance**                          | A child class inherits from **one** parent class.                                                                          | `python class Animal:     def speak(self):         print("Animal sound")  class Dog(Animal):     pass  d = Dog() d.speak() `                                                                  | Used when extending the functionality of one base class (e.g., Car → ElectricCar).               |
| **2. Multilevel Inheritance**                      | A class inherits from a child class, forming a **chain**.                                                                  | `python class Animal:     def eat(self):         print("Eating")  class Mammal(Animal):     pass  class Dog(Mammal):     pass  d = Dog() d.eat() `                                            | Useful in hierarchical structures (e.g., LivingBeing → Animal → Dog).                            |
| **3. Multiple Inheritance**                        | A child class inherits from **multiple parent classes**.                                                                   | `python class Walk:     def walk(self):         print("Walking")  class Swim:     def swim(self):         print("Swimming")  class Duck(Walk, Swim):     pass  d = Duck() d.walk() d.swim() ` | Used when a class needs to combine behaviors from multiple sources (e.g., HybridVehicles).       |
| **4. Hierarchical Inheritance**                    | **Multiple child classes** inherit from the **same parent**.                                                               | `python class Animal:     def sound(self):         print("Animal sound")  class Dog(Animal):     pass  class Cat(Animal):     pass  Dog().sound() Cat().sound() `                             | Best for creating multiple subclasses from one base—e.g., Employee → Manager, Developer, Intern. |
| **5. Hybrid Inheritance**                          | A combination of two or more inheritance types (e.g., multilevel + multiple).                                              | `python class A:     pass  class B(A):     pass  class C(A):     pass  class D(B, C):     pass `                                                                                              | Used in complex systems like frameworks, where classes share and extend functionalities.         |

### Polymorphism
Same Function name but different behaviour or implementation.

| ------------------------------------------------------------------------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------------------------- | -------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- | --------------------------------------------------------------------------------------------------------- |
| **1. Compile-Time Polymorphism (Method Overloading)** *(*Python doesn’t support traditional overloading but allows default arguments / variable-length arguments*)* | Multiple methods with the **same name but different parameters**. Python simulates this using default arguments or `*args`.                 | `python def add(a, b=0, c=0):     return a + b + c  print(add(2))       # 2 print(add(2, 3))     # 5 print(add(2, 3, 4)) # 9 `                                                                                                         | Used to provide the same method name with flexible input (e.g., built-in `len()`, math functions).        |

Runtime Polymorphism (Method Overriding)**                                                                                                                     | A **child class** redefines a method of the **parent class** and is resolved at runtime.                                                    | `python class Animal:     def speak(self):         print("Animal sound")  class Dog(Animal):     def speak(self):         print("Bark")  a = Animal() a.speak()  # Animal sound d = Dog() d.speak()  # Bark `                          | Used in OOP to provide specific behavior for subclasses (e.g., different payment methods, vehicle types). |
